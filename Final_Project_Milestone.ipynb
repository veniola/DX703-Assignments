````python
# =====================
# Inspect the Dataset
# =====================

print("Dataset Inspection:\n")

# --- Number of Samples and Classes
number_samples = len(huff_all)
number_classes = len(set(huff_all['category']))
print(f"Number of samples: {number_samples}")
print(f"Number of classes: {number_classes}")
print()


# --- View Example Records
for i in range(6):
    ex = huff_all[i]
    print(f"Example {i}:")
    print("  category          :", ex.get("category"))
    print("  headline          :", (ex.get("headline") or "")[:120])
    print("  authors           :", (ex.get("authors") or "")[:120])
    print("  link              :", ex.get("link"))
    print("  short_description :", (ex.get("short_description") or "")[:120])
    print("  date              :", ex.get("date"))
    print()
print()


# --- Distribution of Labels (Check for Imbalance)
label_counts = Counter(ex.get("category") for ex in huff_all)
print("Label Distribution:")
for label, count in label_counts.items():
    print(f"  {label}: {count}")
print()


# --- Check for Missing Values or Inconsistent Data
# Missing Values Check
missing_counts = Counter()
for ex in huff_all:
    for key in ['category', 'headline', 'authors', 'link', 'short_description', 'date']:
        if not ex.get(key):
            missing_counts[key] += 1

print("Missing Values:")
for key, count in missing_counts.items():
    print(f"  {key}: {count}")
print()


# --- Duplicate Records Check
duplicate_counts = Counter()
seen = set()
for ex in huff_all:
    identifier = (ex.get("headline"), ex.get("short_description"))
    if identifier in seen:
        duplicate_counts[identifier] += 1
    else:
        seen.add(identifier)

print("Duplicate Records:")
for identifier, count in duplicate_counts.items():
    print(f"  {identifier}: {count}")
print()


# --- Empty Values Check
empty_counts = Counter()
for ex in huff_all:
    for key in ['category', 'headline', 'authors', 'link', 'short_description', 'date']:
        if ex.get(key) == "":
            empty_counts[key] += 1

print("Empty Values:")
for key, count in empty_counts.items():
    print(f"  {key}: {count}")
print()


# --- Overlapping or Ambiguous Records Check
'''
The following are considered overlapping or ambiguous records because they refer to similar content but are labeled differently:
EDUCATION and COLLEGE, 
PARENTS and PARENTING, 
ARTS & CULTURE and ARTS, 
THE WORLDPOST and WORLDPOST, 
TASTE and FOOD & DRINK.
'''
````

````python
# =======================================
# Visualize Key Aspects of the Dataset
# =======================================

# --- For Text, Visualize Class Balance and Linguistic Properties
# Plot number of samples per label (class balance or imbalance)
plt.figure(figsize=(12, 10))
sns.countplot(y=[ex.get("category") for ex in huff_all], order=[label for label, _ in label_counts.most_common()])
plt.title("Class Balance")
plt.show()


# --- Compute Basic Text Statistics
average_word_count = sum(len(word) for ex in huff_all for word in (ex.get("short_description") or "").split()) / max(1, sum(len((ex.get("short_description") or "").split()) for ex in huff_all))
print(f"The Average Word Count in 'Short Descriptions': {average_word_count}")
print()


# --- Examine Examples for Duplicates, Near-Duplicates, or Ambiguous Records
# Ambiguous records based on overlapping content and similar labels
ambiguous_records = ['EDUCATION', 'COLLEGE', 'PARENTS', 'PARENTING', 'ARTS & CULTURE', 'ARTS', 'THE WORLDPOST', 'WORLDPOST', 'TASTE', 'FOOD & DRINK']

# Pull 5 random examples from each ambiguous category group
import random
ambiguous_records_education = random.sample([ex for ex in huff_all if ex.get("category") in ["EDUCATION", "COLLEGE"]], k=min(5, len([ex for ex in huff_all if ex.get("category") in ["EDUCATION", "COLLEGE"]])))
ambiguous_records_parenting = random.sample([ex for ex in huff_all if ex.get("category") in ["PARENTS", "PARENTING"]], k=min(5, len([ex for ex in huff_all if ex.get("category") in ["PARENTS", "PARENTING"]])))
ambiguous_records_arts = random.sample([ex for ex in huff_all if ex.get("category") in ["ARTS & CULTURE", "ARTS"]], k=min(5, len([ex for ex in huff_all if ex.get("category") in ["ARTS & CULTURE", "ARTS"]])))
ambiguous_records_worldpost = random.sample([ex for ex in huff_all if ex.get("category") in ["THE WORLDPOST", "WORLDPOST"]], k=min(5, len([ex for ex in huff_all if ex.get("category") in ["THE WORLDPOST", "WORLDPOST"]])))
ambiguous_records_food = random.sample([ex for ex in huff_all if ex.get("category") in ["TASTE", "FOOD & DRINK"]], k=min(5, len([ex for ex in huff_all if ex.get("category") in ["TASTE", "FOOD & DRINK"]])))

# Display ambiguous records for review
print("Ambiguous Records - Education:")
for ex in ambiguous_records_education:
    print(f"  Headline: {ex.get('headline')} | Short Description: {ex.get('short_description')} | Category: {ex.get('category')}")

print("\nAmbiguous Records - Parenting:")
for ex in ambiguous_records_parenting:
    print(f"  Headline: {ex.get('headline')} | Short Description: {ex.get('short_description')} | Category: {ex.get('category')}")

print("\nAmbiguous Records - Arts:")
for ex in ambiguous_records_arts:
    print(f"  Headline: {ex.get('headline')} | Short Description: {ex.get('short_description')} | Category: {ex.get('category')}")

print("\nAmbiguous Records - Worldpost:")
for ex in ambiguous_records_worldpost:
    print(f"  Headline: {ex.get('headline')} | Short Description: {ex.get('short_description')} | Category: {ex.get('category')}")

print("\nAmbiguous Records - Food:")
for ex in ambiguous_records_food:
    print(f"  Headline: {ex.get('headline')} | Short Description: {ex.get('short_description')} | Category: {ex.get('category')}")
````

## Appendix: A quick guide to Hugging Face Datasets

#### 1) What are they?

* A **table-like** dataset: rows = examples, columns = named fields (e.g., `"image"`, `"label"`, `"headline"`).
* Backed by **Apache Arrow** → fast, memory-efficient, lazy transforms.
* Two core objects:

  * `Dataset` — one table of rows/columns.
  * `DatasetDict` — a dict of splits, e.g. `{"train": Dataset, "val": Dataset, "test": Dataset}`.

from datasets.features import ClassLabel

### 6) Working with Text

You can preprocess and tokenize text datasets using either a **transformer tokenizer** or a **linguistic pipeline like spaCy**, depending on your model type and goals.


#### Option A: Transformer Tokenizer (for fine-tuning models like BERT or DistilBERT)

Use a pretrained tokenizer with the Hugging Face `map` method to efficiently process your dataset in batches:

```python
from transformers import AutoTokenizer

tok = AutoTokenizer.from_pretrained("distilbert-base-uncased")

def tokenize(batch):
    return tok(batch["text"], truncation=True, padding="max_length", max_length=128)

tokenized = huff.map(
    tokenize,
    batched=True,
    remove_columns=[c for c in huff.column_names if c not in ("category", "label")]
)
```

This produces token IDs, attention masks, and other fields expected by transformer models.
Use this approach if your project involves fine-tuning pretrained language models.


#### Option B: spaCy Tokenization and Cleaning (for classical ML or custom preprocessing)

If you are **not using transformers**, or if you want to explore feature engineering (e.g., TF-IDF, word frequency, or embedding averages), you can preprocess text with **spaCy** instead.

```python
# !pip install spacy
# !python -m spacy download en_core_web_sm

import spacy
from datasets import load_dataset

nlp = spacy.load("en_core_web_sm", disable=["ner", "parser", "textcat"])
STOP = spacy.lang.en.stop_words.STOP_WORDS

def spacy_clean(batch):
    docs = list(nlp.pipe(batch["text"], batch_size=1000))
    cleaned = []
    for doc in docs:
        tokens = [t.lemma_.lower() for t in doc if t.is_alpha and not t.is_stop]
        cleaned.append(" ".join(tokens))
    return {"text_clean": cleaned, "len_tokens": [len(c.split()) for c in cleaned]}

huff = huff.map(
    spacy_clean,
    batched=True,
    batch_size=1000,
    remove_columns=[c for c in huff.column_names if c not in ("category", "label")]
)
```

The resulting column `text_clean` can be used with:

* `TfidfVectorizer` (Scikit-learn)
* `TextVectorization` (Keras)
* or any other custom embedding method.

> 💡 **Tip:** spaCy is ideal for lightweight NLP pipelines or for models that rely on explicit preprocessing (lemmatization, stopword removal). Transformer tokenizers, by contrast, expect *raw text* and handle subword tokenization internally.

#### 8) Quick “gotchas”

* **Columns by name** (strings), not numeric indices.
* Avoid converting huge columns to `list(...)` unless necessary; prefer vectorized ops with `map`, `filter`, `select`.
* `PYTHONHASHSEED` must be set **before** the Python process starts to matter; use explicit `seed=` arguments for reproducibility.
* Datasets print **previews** (e.g., `Column([6, 6, 6, ...])` is just the first few values).


#### 9) A minimal checklist to follow

1. `load_dataset(...)` → confirm `column_names`, `features`.
2. Build any needed columns (`"text"`), and **class-encode** labels if strings.
3. Make a **stratified 80/10/10** split (`train_test_split` ×2).
4. Do **EDA**: class counts, sample printouts or image grids.
5. For training:

   * **images** → `with_transform` + DataLoader
   * **text** → tokenizer via `.map(...)` + trainer/model pipeline
6. Save your `DatasetDict` with `save_to_disk(...)` (optional for text but handy, **don't** use it for big image datasets).
